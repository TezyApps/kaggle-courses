### Mathematical Transformations:

```python
from IPython.display import display

import pandas as pd
import numpy as np
```

```python
autos = pd.read_csv('../../resources/autos.csv')
autos.head()
```

```python
columns = [col for col in autos.columns]
print(columns)
```

```python
autos[["stroke", "bore"]].head()
```

With pandas built-in functions, it's quite useful to create custom computed transformations that makes more meaningful features that contributes to machine learning

For example:
1. Calculating `stroke_ratio` from `stroke` and `bore` data
2. Calculating `displacement`, etc.

```python
# Create a new feature by computing two existing features
autos['stroke_ratio'] = autos['stroke'] / autos['bore']

autos[['stroke', 'bore', 'stroke_ratio']].head()
```

```python
# calculating displament
autos['displacement'] = (
    np.pi * ((0.5 * autos['bore']) ** 2) * autos['stroke'] * autos['num_of_cylinders']
)

autos['displacement'].head()
```

### Counts

```python
sample_data = {
    "amenity": [True, False, False, False],
    "bump": [False, False, False, False],
    "crossing": [True, True, False, False],
    "emergency": [False, False, False, True]
}

df = pd.DataFrame(sample_data)
df.head()
```

#### Computed counts by Axis
- **0** - Columns
- **1** - Rows

```python
df['overview'] = df.sum(axis=1)
df.head()
```

```python
df['valid'] = df.sum(axis=1).gt(1)
df.head()
```

#### Building Up & Breaking Down

```python
policy = {
    "policy": ["corp l2", "corp l3", "corp l4", "corp l5", "per l1","per l2", "per l3", "per l4", "per l5"],
}
df = pd.DataFrame(policy)
df.head()
```

```python
# split the policy column into two columns
df[['type', 'level']] = (
    df['policy'].str.split(' ', expand=True)
)
print(df)
```

```python
# merge two columns' values into one column
df['access'] = (
    (df['type'] + '_' + df['level']).str.upper()
)
df.head()
```

### Group Transforms

```python
income_by_states = {
    "country": ["Australia", "Australia", "Australia", "Australia", "Australia", "Australia", "Australia", "Australia"],
    "state": ["ACT", "NT", "WA", "NSW", "VIC", "QLD", "SA", "TAS"],
    "income": [75643, 66831, 62207, 58909, 57907, 56708, 55782, 53479]
}
df = pd.DataFrame(income_by_states)
df.head()
```

```python
df['avg_income'] = (
    df.groupby('country')
    ['income']
    .transform('mean')
)

display(df)
```

```python
display(df['avg_income'].unique())
```

```python
attendance = {
    "check_in": ["in", "in", "in", "out", "in", "out", "out", "out", "in" ]
}
df = pd.DataFrame(attendance)
display(df)
```

```python
df['check_in_count'] = (
    df.groupby('check_in')
    ['check_in']
    .transform('count')
)
print(df)
```

```python
df['check_in_frequency'] = (
    df['check_in_count'] / df['check_in'].count()
)
display(df)
```

```python
df['check_in_factorize'], _ = df['check_in_count'].factorize()
display(df)
```

```python
display(df.value_counts())
```

```python
from sklearn.model_selection import cross_val_score
from xgboost import XGBRegressor
```

```python
def score_dataset(X, y, model=XGBRegressor()):
    # Label encoding for categoricals
    for colname in X.select_dtypes(["category", "object"]):
        X[colname], _ = X[colname].factorize()
    # Metric for Housing competition is RMSLE (Root Mean Squared Log Error)
    score = cross_val_score(
        model, X, y, cv=5, scoring="neg_mean_squared_log_error",
    )
    score = -1 * score.mean()
    score = np.sqrt(score)
    return score
```
